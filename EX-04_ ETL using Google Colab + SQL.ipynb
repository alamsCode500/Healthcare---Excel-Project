# **Transform Phase**

**Standardize Date Format**

```python
# Standardize Date Format

pd.read_sql("""
SELECT
    DATE(Date) AS date,
    "State/UnionTerritory" AS state,
    Confirmed,
    Deaths
FROM covid_cases
""", conn)
```

**Daily New Cases**

```python
# Daily New Cases

pd.read_sql("""
SELECT
    state,
    date,
    Confirmed -
    LAG(Confirmed) OVER (PARTITION BY state ORDER BY date)
    AS Daily_New_Cases
FROM (
    SELECT
        DATE(Date) AS date,
        "State/UnionTerritory" AS state,
        Confirmed
    FROM covid_cases
)
""", conn)
```

**Total Confirmed (Indian + Foreign)**

```python
# Total Confirmed (Indian + Foreign)

pd.read_sql("""
SELECT
    state,
    date,
    (ConfirmedIndianNational + ConfirmedForeignNational) AS Total_Confirmed
FROM (
    SELECT
        DATE(Date) AS date,
        "State/UnionTerritory" AS state,
        ConfirmedIndianNational,
        ConfirmedForeignNational
    FROM covid_cases
)
""", conn)
```

**Case Fatality Rate**

```python
# Case Fatality Rate

pd.read_sql("""
SELECT
    state,
    date,
    ROUND((Deaths * 100.0 / Confirmed), 2) AS CFR
FROM (
    SELECT
        DATE(Date) AS date,
        "State/UnionTerritory" AS state,
        Confirmed,
        Deaths
    FROM covid_cases
)
WHERE Confirmed > 0
""", conn)
```

**Join Cases + Testing**

```python
# Join Cases + Testing

pd.read_sql("""
SELECT
    c.state,
    c.date,
    c.Confirmed,
    t.TotalSamples,
    t.Positive
FROM (
    SELECT DATE(Date) AS date,
           "State/UnionTerritory" AS state,
           Confirmed
    FROM covid_cases
) c
JOIN covid_testing t
ON c.state = t.State AND c.date = DATE(t.Date)
""", conn)
```

# **Load Phase**

**Drop existing table**

```python
# Drop existing table
conn.execute("DROP TABLE IF EXISTS covid_summary")
conn.commit()

conn.execute("DROP TABLE IF EXISTS vacc_clean")
conn.commit()
```

**Create Table vacc_clean**

```python
# Create Table vacc_clean

conn.execute("""
CREATE TABLE vacc_clean AS
SELECT
    UPPER(TRIM(State)) AS state,

    -- Convert DD/MM/YYYY → YYYY-MM-DD
    SUBSTR("Updated On", 7, 4) || '-' ||
    SUBSTR("Updated On", 4, 2) || '-' ||
    SUBSTR("Updated On", 1, 2) AS date,

    MAX(COALESCE("First Dose Administered", 0))  AS First_Dose,
    MAX(COALESCE("Second Dose Administered", 0)) AS Second_Dose

FROM covid_vaccination
GROUP BY state, date;

""")

conn.commit()
```

**Verify table vacc_clean**

```python
# Verify table vacc_clean

pd.read_sql("SELECT * FROM vacc_clean", conn)
```

**Create Table covid_summary**

```python
# Create Table covid_summary

conn.execute("""
CREATE TABLE covid_summary AS
SELECT
    c.state,
    c.date,
    c.Confirmed,
    c.Deaths,
    c.Cured,
    c.Daily_New_Cases,
    t.TotalSamples,
    t.Positive,

    COALESCE(v.First_Dose, 0)  AS First_Dose,
    COALESCE(v.Second_Dose, 0) AS Second_Dose,

    ROUND(
        CASE WHEN c.Confirmed > 0
        THEN (c.Deaths * 100.0 / c.Confirmed)
        ELSE 0 END, 2
    ) AS CFR,

    ROUND(
        CASE WHEN t.TotalSamples > 0
        THEN (t.Positive * 100.0 / t.TotalSamples)
        ELSE 0 END, 2
    ) AS Positive_Rate

FROM (
    SELECT
        DATE(Date) AS date,
        UPPER(TRIM("State/UnionTerritory")) AS state,
        Confirmed,
        Deaths,
        Cured,
        Confirmed -
        LAG(Confirmed) OVER (
            PARTITION BY UPPER(TRIM("State/UnionTerritory"))
            ORDER BY DATE(Date)
        ) AS Daily_New_Cases
    FROM covid_cases
) c

LEFT JOIN covid_testing t
    ON UPPER(TRIM(t.State)) = c.state
   AND DATE(t.Date) = c.date

LEFT JOIN vacc_clean v
    ON v.state = c.state
   AND v.date = c.date;
""")

conn.commit()
```

**Verify table covid_summary**

```python
# Verify table covid_summary

pd.read_sql("SELECT * FROM covid_summary", conn)
```

**Risk Level Column**

```python
# Risk Level Column

final_df = pd.read_sql("""
SELECT *,
CASE
    WHEN CFR > 2 AND Positive_Rate > 10 THEN 'High'
    WHEN CFR > 2 OR Positive_Rate > 10 THEN 'Medium'
    ELSE 'Low'
END AS Risk_Level
FROM covid_summary
""", conn)

final_df
```
